```python
# Extract F1, MCC, EV values for LaTeX table
from occhio.benchmark.configs import BenchmarkName

sae_labels = ["Standard", "Matryoshka", "MatchingPursuit"]
distributions = list(BenchmarkName)

with open("/tmp/sae_results.txt", "w") as f:
    for model_idx, benchmark in enumerate(distributions):
        model = grid[model_idx]
        s = f"{benchmark.value}: "
        for sae_label in sae_labels:
            f1 = model.saes_f1_score[sae_label]
            mcc = model.saes_mcc[sae_label]
            ev = model.saes_explained_variance[sae_label]
            s += f"{sae_label}(F={f1:.2f},M={mcc:.2f},E={ev:.2f}) "
        f.write(s + "\n")
print("Saved to /tmp/sae_results.txt")
```
